```python
#11
s1=input()
s2=input()
s1vals=[]
s2vals=[]
s1len=len(s1)
s2len=len(s2)
count=0
for s in s1:
  s1vals.append(s)
for s in s2:
  s2vals.append(s)
for s in s1vals:
    if all(x in s2vals for x in s1vals)==True:
      count+=1
if(count==s1len or count==s2len) and (s1len==s2len):
  print("Anagram")
else:
  print("Not anagram")
```

```python
#16
s=input()
v="aeiou"
pos=[]
vol=[]
for i in range(len(s)):
  if s[i] in v:
    pos.append(i)
    vol.append(s[i])
vol.reverse()
s=list(s)
for i in range(len(pos)):
  s[pos[i]] = vol[i]
print("".join(s))
```
